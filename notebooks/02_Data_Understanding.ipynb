```python
import pandas as pd
```

```python
Customers = pd.read_csv("../data/Customers.csv")
Orders = pd.read_csv("../data/Orders.csv")
Order_Details = pd.read_csv("../data/Order_Details.csv")
Products = pd.read_csv("../data/Products.csv")
Targets = pd.read_csv("../data/Targets.csv")
```

Cheak Column Names

```python
print("Customers Columns:", Customers.columns.tolist())
print("Products Columns:", Products.columns.tolist())
print("Orders Columns:", Orders.columns.tolist())
print("Order Details Columns:", Order_Details.columns.tolist())
print("Targets Columns:", Targets.columns.tolist())
```

Check Missing Values

```python
print("CUSTOMERS")
display(Customers.isnull().sum())

print("PRODUCTS")
display(Products.isnull().sum())

print("ORDERS")
display(Orders.isnull().sum())

print("ORDER DETAILS")
display(Order_Details.isnull().sum())

print("TARGETS")
display(Targets.isnull().sum())
```

Duplicate Records

```python
print("Customers Duplicates:", Customers.duplicated().sum())
print("Products Duplicates:",Products.duplicated().sum())
print("Orders Duplicates:",Orders.duplicated().sum())
print("Order Details Duplicates:",Order_Details.duplicated().sum() )
print("Targets Duplicates:",Targets.duplicated().sum())
```

DATA SUMMARY

```python
data_summary = pd.DataFrame({
    "Dataset": [
        "Customers",
        "Products",
        "Orders",
        "Order_Details",
        "Targets"
    ],
    "Rows": [
        Customers.shape[0],
        Products.shape[0],
        Orders.shape[0],
        Order_Details.shape[0],
        Targets.shape[0]
    ],
    "Columns": [
        Customers.shape[1],
        Products.shape[1],
        Orders.shape[1],
        Order_Details.shape[1],
        Targets.shape[1]
    ],
    "Duplicates": [
        Customers.duplicated().sum(),
        Products.duplicated().sum(),
        Orders.duplicated().sum(),
        Order_Details.duplicated().sum(),
        Targets.duplicated().sum()
    ]
})

display(data_summary)
```
