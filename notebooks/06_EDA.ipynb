Step 10: Monthly Sales Trends

```python
sales_data["Month"] = sales_data["Order_Date"].dt.month
```

```python
sales_data.columns
```

Create Monthly Sales Analysis

Create Year-Month Column

```python
sales_data["Order_Date"] = pd.to_datetime(
    sales_data["Order_Date"],
    errors="coerce"
)

sales_data["Year_Month"] = (
    sales_data["Order_Date"].dt.to_period("M")
)
```

Monthly Sales Analysis

```python
monthly_sales = (
    sales_data.groupby("Year_Month")["Net_Sales"]
    .sum()
    .sort_index()
)

monthly_sales
```

Monthly Sales DataFrame

```python
monthly_sales_df = (
    sales_data.groupby("Year_Month")
    .agg(
        Total_Sales=("Net_Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Orders=("Order_ID", "nunique")
    )
    .reset_index()
)

monthly_sales_df
```

Step 12: Business Insights Summary

Regional Sales

```python
region_summary = (
    sales_data.groupby("Region")
    .agg(
        Total_Sales=("Net_Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Orders=("Order_ID", "nunique")
    )
    .reset_index()
)

region_summary
```

Category Sales

```python
category_summary = (
    sales_data.groupby("Category_Details")
    .agg(
        Total_Sales=("Net_Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)

category_summary
```
