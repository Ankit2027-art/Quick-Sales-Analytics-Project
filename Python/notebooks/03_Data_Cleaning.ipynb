Step 13: Clean Text Columns

Why? Extra spaces can cause problems when filtering, grouping, or joining data.

```python
# Remove leading and trailing spaces from text columns

dataframes = [
    Customers_clean,
    Products_clean,
    Orders_clean,
    Order_details_clean,
    Targets_clean
]

for df in dataframes:
    text_columns = df.select_dtypes(include="object").columns

    for column in text_columns:
        df[column] = df[column].str.strip()

print("Text columns cleaned successfully")
print("Text columns cleaned successfully")
```

Verify Text Cleaning

```python
# Check for leading or trailing spaces

for name, df in {
    "Customers": Customers_clean,
    "Products": Products_clean,
    "Orders": Orders_clean,
    "Order Details": Order_details_clean,
    "Targets": Targets_clean
}.items():

    text_columns = df.select_dtypes(include="object").columns
    space_issues = 0

    for column in text_columns:
        space_issues += (
            df[column].dropna() != df[column].dropna().str.strip()
        ).sum()

    print(f"{name} space issues: {space_issues}")
```

Step 13: Validate Business Rules

verify:

Gross_Sales = Quantity × Selling_Price

Discount_Amount = Gross_Sales × Discount

Net_Sales = Gross_Sales − Discount_Amount

Total_Cost = Quantity × Cost_Price

Profit = Net_Sales − Total_Cost

Check Gross Sales

```python
# Calculate expected Gross Sales

expected_gross_sales = (
    Order_details_clean["Quantity"]
    * Order_details_clean["Selling_Price"]
)

gross_sales_errors = (
    (Order_details_clean["Gross_Sales"] - expected_gross_sales).abs() > 0.01
).sum()

print("Gross Sales Errors:", gross_sales_errors)
```

Step 14: Validate Discount Amount

```python
# Calculate expected Discount Amount

expected_discount_amount = (
    Order_details_clean["Gross_Sales"]
    * Order_details_clean["Discount"]
)

discount_errors = (
    (
        Order_details_clean["Discount_Amount"]
        - expected_discount_amount
    ).abs() > 0.01
).sum()

print("Discount Amount Errors:", discount_errors)
```

Step 15: Validate Net Sales

```python
# Calculate expected Net Sales

expected_net_sales = (
    Order_details_clean["Gross_Sales"]
    - Order_details_clean["Discount_Amount"]
)

net_sales_errors = (
    (
        Order_details_clean["Net_Sales"]
        - expected_net_sales
    ).abs() > 0.01
).sum()

print("Net Sales Errors:", net_sales_errors)
```

Step 16: Validate Total Cost

```python
# Calculate expected Total Cost

expected_total_cost = (
    Order_details_clean["Quantity"]
    * Order_details_clean["Cost_Price"]
)

total_cost_errors = (
    (
        Order_details_clean["Total_Cost"]
        - expected_total_cost
    ).abs() > 0.01
).sum()

print("Total Cost Errors:", total_cost_errors)
```

Step 17: Validate Profit
Formula:

Profit = Net_Sales − Total_Cost

```python
# Calculate expected Profit

expected_profit = (
    Order_details_clean["Net_Sales"]
    - Order_details_clean["Total_Cost"]
)

profit_errors = (
    (
        Order_details_clean["Profit"]
        - expected_profit
    ).abs() > 0.01
).sum()

print("Profit Errors:", profit_errors)
```

Step 18: Validate Profit Margin

Formula   :

Profit Margin = (Profit/Net Sales)*100

```python
# Check Profit Margin values

display(
    Order_details_clean[
        ["Net_Sales", "Profit", "Profit_Margin"]
    ].head(10)
)
```

Step 18: Validate Profit Margin

```python
# Calculate expected Profit Margin

expected_profit_margin = (
    Order_details_clean["Profit"]
    / Order_details_clean["Net_Sales"]
) * 100

# Handle zero Net Sales
expected_profit_margin = expected_profit_margin.fillna(0)

display(
    pd.DataFrame({
        "Existing Profit Margin": Order_details_clean["Profit_Margin"].head(10),
        "Expected Profit Margin": expected_profit_margin.head(10)
    })
)
```

Check Profit Margin Errors

```python
# Validate Profit Margin

profit_margin_errors = (
    (
        Order_details_clean["Profit_Margin"]
        - expected_profit_margin
    ).abs() > 0.01
).sum()

print("Profit Margin Errors:", profit_margin_errors)
```

Compare Existing and Calculated Values

```python
# Inspect Profit Margin values

margin_check = pd.DataFrame({
    "Net_Sales": Order_details_clean["Net_Sales"],
    "Profit": Order_details_clean["Profit"],
    "Existing_Margin": Order_details_clean["Profit_Margin"],
    "Calculated_Margin": expected_profit_margin
})

display(margin_check.head(10))
```

Check Profit Margin Statistics

```python
# Check Profit Margin range

print("Existing Profit Margin:")
print(Order_details_clean["Profit_Margin"].describe())

print("\nCalculated Profit Margin:")
print(expected_profit_margin.describe())
```

Check One Row Manually

```python
# Manually calculate Profit Margin for the first row

row = Order_details_clean.iloc[0]

manual_margin = (
    row["Profit"] / row["Net_Sales"]
) * 100

print("Existing Margin:", row["Profit_Margin"])
print("Calculated Margin:", manual_margin)
```

##  Problem

Profit_Margin values are stored as decimal values, not percentages.

Step 19: Correct Profit Margin Validation

```python
# Calculate expected Profit Margin as a decimal

expected_profit_margin_decimal = (
    Order_details_clean["Profit"]
    / Order_details_clean["Net_Sales"]
).fillna(0)

# Compare with existing Profit_Margin

profit_margin_errors = (
    (
        Order_details_clean["Profit_Margin"]
        - expected_profit_margin_decimal
    ).abs() > 0.01
).sum()

print("Profit Margin Errors:", profit_margin_errors)
```
